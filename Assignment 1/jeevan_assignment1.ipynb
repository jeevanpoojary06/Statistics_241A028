### Remove Outliers Using IQR

```python
# Copying the dataset

df_processed = df.copy()


for column in numerical_columns:
    
    Q1 = df_processed[column].quantile(0.25)
    
    Q3 = df_processed[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    
    lower_limit = Q1 - (1.5*IQR)
    
    upper_limit = Q3 + (1.5*IQR)
    
    
    df_processed = df_processed[
        
        (df_processed[column] >= lower_limit)
        &
        (df_processed[column] <= upper_limit)
        
    ]
```

"Outliers were identified and removed using the Interquartile Range (IQR) method as part of the preprocessing pipeline. Although some extreme values may represent legitimate observations, the IQR method was employed to reduce the influence of unusually large or small values on subsequent analysis and preprocessing steps."
